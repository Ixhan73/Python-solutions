## 8-7. Album

```python
def make_album(artist, title, tracks=None):
    album = {
        "artist": artist,
        "title": title
    }

    if tracks is not None:
        album["tracks"] = tracks

    return album


album_1 = make_album("Pink Floyd", "The Dark Side of the Moon")
album_2 = make_album("Michael Jackson", "Thriller")
album_3 = make_album("Adele", "25")
album_4 = make_album("Linkin Park", "Meteora", 13)

print(album_1)
print(album_2)
print(album_3)
print(album_4)
```

## 8-8. User Albums

```python
def make_album(artist, title, tracks=None):
    album = {
        "artist": artist,
        "title": title
    }

    if tracks is not None:
        album["tracks"] = tracks

    return album


while True:
    print("\nEnter album information (or enter 'q' to quit).")
    artist = input("Artist: ")

    if artist.lower() == "q":
        break

    title = input("Album title: ")

    if title.lower() == "q":
        break

    print(make_album(artist, title))
```

## 8-9. Magicians

```python
def show_magicians(magicians):
    for magician in magicians:
        print(magician)


magicians = ["David Copperfield", "Penn Jillette", "Teller"]

show_magicians(magicians)
```

## 8-10. Great Magicians

```python
def show_magicians(magicians):
    for magician in magicians:
        print(magician)


def make_great(magicians):
    for index in range(len(magicians)):
        magicians[index] = f"{magicians[index]} the Great"


magicians = ["David Copperfield", "Penn Jillette", "Teller"]

make_great(magicians)
show_magicians(magicians)
```

## 8-11. Unchanged Magicians

```python
def show_magicians(magicians):
    for magician in magicians:
        print(magician)


def make_great(magicians):
    great_magicians = []

    for magician in magicians:
        great_magicians.append(f"{magician} the Great")

    return great_magicians


magicians = ["David Copperfield", "Penn Jillette", "Teller"]
great_magicians = make_great(magicians.copy())

print("Original list:")
show_magicians(magicians)

print("\nGreat magicians:")
show_magicians(great_magicians)
```

## 8-12. Sandwiches

```python
def make_sandwich(*items):
    print("\nSandwich ordered with:")

    for item in items:
        print(f"- {item}")


make_sandwich("chicken")
make_sandwich("chicken", "cheese", "lettuce")
make_sandwich("beef", "cheese", "tomato", "onion", "sauce")
```

## 8-13. User Profile

```python
def build_profile(first, last, **user_info):
    user_info["first_name"] = first
    user_info["last_name"] = last

    return user_info


profile = build_profile(
    "Izhan",
    "Ahmad",
    field="Artificial Intelligence",
    language="Python",
    goal="AI Engineer"
)

print(profile)
```

## 8-14. Cars

```python
def make_car(manufacturer, model, **car_info):
    car_info["manufacturer"] = manufacturer
    car_info["model"] = model

    return car_info


car = make_car(
    "subaru",
    "outback",
    color="blue",
    tow_package=True
)

print(car)
```

## 8-17. Styling Functions
