## 6-5. Rivers

```python
rivers = {
    "nile": "egypt",
    "amazon": "brazil",
    "indus": "pakistan"
}

for river, country in rivers.items():
    print(f"The {river.title()} runs through {country.title()}.")

print("\nRivers:")
for river in rivers:
    print(river.title())

print("\nCountries:")
for country in rivers.values():
    print(country.title())
```

## 6-6. Polling

```python
favorite_languages = {
    "jen": "python",
    "sarah": "c",
    "edward": "rust",
    "phil": "python"
}

people_to_poll = ["jen", "sarah", "mike", "phil", "ahmad"]

for person in people_to_poll:
    if person in favorite_languages:
        print(f"Thank you, {person.title()}, for responding to the poll.")
    else:
        print(f"{person.title()}, please take our favorite languages poll.")
```

## 6-11. Cities

```python
cities = {
    "peshawar": {
        "country": "Pakistan",
        "population": "2.3 million",
        "fact": "It is one of the oldest cities in South Asia."
    },
    "london": {
        "country": "United Kingdom",
        "population": "9 million",
        "fact": "It is home to the River Thames."
    },
    "tokyo": {
        "country": "Japan",
        "population": "14 million",
        "fact": "It is one of the world's most populous metropolitan areas."
    }
}

for city, information in cities.items():
    print(f"City: {city.title()}")
    print(f"Country: {information['country']}")
    print(f"Population: {information['population']}")
    print(f"Fact: {information['fact']}\n")
```

## 6-12. Extensions

```python
cities = {
    "peshawar": {
        "country": "Pakistan",
        "population": "2.3 million",
        "fact": "It is one of the oldest cities in South Asia.",
        "language": "Pashto and Urdu",
        "famous_for": "Qissa Khwani Bazaar"
    },
    "london": {
        "country": "United Kingdom",
        "population": "9 million",
        "fact": "It is home to the River Thames.",
        "language": "English",
        "famous_for": "Big Ben"
    },
    "tokyo": {
        "country": "Japan",
        "population": "14 million",
        "fact": "It is one of the world's most populous metropolitan areas.",
        "language": "Japanese",
        "famous_for": "Technology and modern city life"
    }
}

for city, information in cities.items():
    print(f"--- {city.title()} ---")
    for key, value in information.items():
        print(f"{key.replace('_', ' ').title()}: {value}")
    print()
```
