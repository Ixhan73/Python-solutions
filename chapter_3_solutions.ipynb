## 3-6. More Guests

Add three more guests using insert() at the beginning, insert() in the middle, and append() at the end.

```python
guests = ["Albert Einstein", "Isaac Newton", "Nikola Tesla"]

print("I found a bigger dinner table, so I can invite more guests.")

guests.insert(0, "Leonardo da Vinci")
guests.insert(2, "Galileo Galilei")
guests.append("Marie Curie")

for guest in guests:
    print(f"{guest}, I would like to invite you to dinner.")
```

## 3-7. Shrinking Guest List

Reduce the guest list to two people using pop(), then remove the remaining guests with del.

```python
guests = [
    "Leonardo da Vinci",
    "Albert Einstein",
    "Galileo Galilei",
    "Isaac Newton",
    "Nikola Tesla",
    "Marie Curie"
]

print("I can invite only two people for dinner.")

while len(guests) > 2:
    removed_guest = guests.pop()
    print(f"Sorry {removed_guest}, I can't invite you to dinner.")

for guest in guests:
    print(f"{guest}, you are still invited to dinner.")

del guests[1]
del guests[0]

print(guests)
```

## 3-8. Seeing the World

Create a list of places you want to visit and demonstrate sorted(), reverse(), and sort() without losing track of the original order.

```python
places = ["Japan", "Iceland", "Turkey", "Switzerland", "New Zealand"]

print("Original order:")
print(places)

print("\nAlphabetical order:")
print(sorted(places))

print("\nOriginal order:")
print(places)

print("\nReverse alphabetical order:")
print(sorted(places, reverse=True))

print("\nOriginal order:")
print(places)

places.reverse()
print("\nReversed order:")
print(places)

places.reverse()
print("\nBack to original order:")
print(places)

places.sort()
print("\nAlphabetical order:")
print(places)

places.sort(reverse=True)
print("\nReverse alphabetical order:")
print(places)
```

## 3-10. Every Function

Create a list and use each list operation/function introduced in this chapter at least once.

```python
cities = ["Peshawar", "Lahore", "Karachi", "Islamabad"]

# Access an element
print(cities[0])

# Change an element
cities[0] = "Quetta"
print(cities)

# Add to the end
cities.append("Multan")
print(cities)

# Insert an element
cities.insert(1, "Peshawar")
print(cities)

# Remove by value
cities.remove("Multan")
print(cities)

# Pop an element
removed_city = cities.pop()
print(removed_city)
print(cities)

# Sort alphabetically
cities.sort()
print(cities)

# Sort in reverse alphabetical order
cities.sort(reverse=True)
print(cities)

# Reverse the current order
cities.reverse()
print(cities)

# Find the length
print(len(cities))

# Delete an element
del cities[0]
print(cities)
```
